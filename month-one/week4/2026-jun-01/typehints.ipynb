# Types Hints for variables


### without types hints


```python
name ="amaldas"
age = 22
print(name,age)
```

### with types hints...............................

#### typehints tells the reader and tools what type of data is expected


```python
name:str ="amaldas"
age:int = 22  
print(name,age)
```

```python
def introduce(name:str,age:int)->str:
    return f"my name is{name} and my age is {age}"
introduce("amal",21)
```

# <b>union</b>

### THIS VALUE CAN BE ONE OF MULTIPLE TYPES

```python
def display(value:str|int)->None:
    print(f"the value is {value}")
display("hello")
display(22)
```

```python
def print_items(items: list[int] | tuple[int, ...]) -> None:
    for item in items:
        print(item)




print_items([1, 2, 3])
print_items((4, 5, 6))
```

# OPTIONAL


### THIS VALUE CAN BE NORMAL TYPE OR NONE

```python
from typing import Optional
def greet(name: Optional[str] = None) -> str:
    if name:
        return f"Hello, {name}!"
    else:
        return "Hello, stranger!"
print(greet("amaldaskk"))
print(greet())
```

# GENERICS.............................................


### GENERICS ALLOWS YOU TO WRITE FUNCTIONS OR CLASS THAT WORKS WITH MANY TYPES WHILE KEEPING TYPE SAFETY

##### WITHOUT GENERICS .............

```python
def get_first(items: list) -> object:
    return items[0]
```

#### The return type is just object, so mypy doesn't know the exact type.

## with generics


```python
from typing import TypeVar
T = TypeVar('T')
def get_values(items: list[T]) -> T:
    return items[0]
name = get_values(["amal", "kumar", "das"])
print(name)
number = get_values([1, 2, 3])
print(number)
```

# Generic class Example........................................................

```python
from typing import TypeVar,Generic
T = TypeVar('T')


class box(Generic[T]):
    def __init__(self,value: T):
        self.value = value
    def get_value(self) -> T:
        return self.value
name = box("amal")
print(name.get_value())
number = box(123)
print(type(number.get_value()))
```
